## Trigonometric Simplification and Transformation in MyEq

In this section we apply trigonometric simplification and transformations
to a symbolic `MyEq` expression.

The example starts from a trigonometric expression assumed equal to zero,
which is useful in optimization problems (maxima and minima), where the
main factor of the expression is relevant.

For example:

```python
P = MyEq(((sin(alpha) - cos(alpha))**2 + (sin(alpha) + cos(alpha))**2) /
         (csc(alpha) * tan(alpha)), 'P')
```

### Trigonometric Simplification

The `tsimplify()` method performs algebraic and trigonometric
simplification of the expression.

```python
P.tsimplify()
```

```python
P=MyEq((-2*sin(alpha)*cos(alpha)+sin(alpha)+cos(alpha)**2+1)*cos(alpha),'P')
```

 

$$
P = (-2\sin(\alpha)\cos(\alpha) + \sin(\alpha) + \cos^2(\alpha) + 1)\cos(\alpha)
$$

### Algebraic Expansion

The `expand()` method performs standard algebraic expansion.

```python
P.expand()
```

produces:

$$
P = -2\sin(\alpha)\cos^2(\alpha) + \sin(\alpha)\cos(\alpha) + \cos^3(\alpha) + \cos(\alpha)
$$

### Convert Sine to Cosine

The `sin2cos(ang)` method rewrites sine powers using cosine identities.

```python
P.sin2cos(alpha)
```

### Convert Cosine to Sine

The `cos2sin(ang)` method rewrites cosine powers using sine identities.

```python
P.cos2sin(alpha)
```

### Trigonometric Expansion

The `texpand()` method expands trigonometric arguments
(e.g., angle-doubling identities).

For example:

```python
P = MyEq(sin(2*alpha), 'P')
```

```python
P.texpand()
```

produces:

$$
P = 2\sin(\alpha)\cos(\alpha)
$$
