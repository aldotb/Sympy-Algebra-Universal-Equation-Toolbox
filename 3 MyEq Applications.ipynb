# MyEq Power and Exponential Operations

In this tutorial we will explore the main power and exponential
transformations available in the `MyEq` class.

We will see how the following methods modify and reorganize the
symbolic expression stored inside a `MyEq` object:

- `Add` — adds an expression to the current MyEq expression  
- `Subs` — substitutes a symbolic part of the expression  
- `Mul` — multiplies the expression by another factor  
- `Pow` — raises the expression to a power  
- `Rpow` — applies a root (inverse power) to the expression  
- `rsimplify` — simplifies root–power combinations  
- `joinbase` — combines powers that share the same base  

Each operation updates the internal symbolic structure while preserving
algebraic equivalence.

The following sections demonstrate these transformations step by step
using practical MyEq examples.

## Symbolic vs In-Place Operations in MyEq

MyEq supports two ways of applying algebraic operations:

- **Symbolic operators** (`+`, `-`, `*`, `**`)  
  These create new symbolic expressions without modifying the original object.

- **In-place methods** (`Add`, `Subs`, `Mul`, `Pow`, `Rpow`)  
  These update the internal expression stored in the MyEq instance.

This distinction allows MyEq objects to behave similarly to symbolic variables while still supporting controlled mutation when explicitly requested.

In the examples below:

```python
A=MyEq(x+2,'A')
B=MyEq(y-4,'B')
```

- `A + 3` and `A + B` produce new expressions but do not change `A` or `B`.
- `A.Add(z)` modifies the internal expression of `A`.
- `A.Pow(B)` raises the current expression of `A` to the power `B` in place.

## Power of a MyEq Expression

If `P` represents an expression ( f(x) ), applying `P.Pow(n)` raises the expression to the power ( n ).

The `Pow` method modifies the internal symbolic expression stored in the `MyEq` object.

After exponentiation, standard algebraic operations such as expansion can be applied to the result.

For example, if:

```python
P=MyEq(x**2-1,'P')
```

then:

```python
P.Pow(2)
```

The result can then be expanded using:

```python
P.expand()
```

## Root Power of a MyEq Expression

If `P` represents an expression raised to a power, applying `P.Rpow(n)`
applies the inverse power (root) to the expression.

$$
P = f(x)^n \Rightarrow P = \sqrt[n]{f(x)^n}
$$

For example, if:

```python
P=MyEq((3+x)**2,'P')
```

then:

```python
P.Rpow(2)
```

produces:

$$
P = \sqrt{(3 + x)^2}
$$

The result can then be simplified using `rsimplify()`.

## Root Simplification of a MyEq Expression

The `rsimplify()` method simplifies expressions where a root is applied
to a power of the same base.

$$
\sqrt[n]{f(x)^n} \Rightarrow f(x)
$$

For example:

```python
P.rsimplify()
```

## Base Combination in a MyEq Expression

The `joinbase()` method combines powers that share the same symbolic base
within multiplicative expressions.

$$
x^m \cdot x^n \Rightarrow x^{m+n}
$$

For example, if:

```python
P=MyEq(x**3*x**a+y**z*(y*x)**b,'P')
```
