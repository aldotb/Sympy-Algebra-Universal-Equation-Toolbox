 

### Philosophy

I think **SymPy** is the best math library for Python, but unfortunately it almost always does what it wants and rearranges expressions as it sees fit, and sometimes it's not so much the result but the **form** that matters.

To avoid this stress, we handle mathematical expressions in a more controlled way:

$$
\text{sadd}(a,b) = a + b
$$

$$
\text{smul}(a,b) = a \cdot b
$$

$$
\text{sdiv}(a,b) = \frac{a}{b}
$$

$$
\text{spow}(a,b) = a^b
$$

$$
\text{sroot}(a,n) = \sqrt[n]{a}
$$

$$
\text{sinverse}(a) = \frac{1}{a}
$$

 
# 📘 Libaldomath   Keywords Guide

## 🔹 Classic Algebra

Operaciones básicas de álgebra simbólica.

| Keyword | Description |
|------------------ | -----------------------------------  |
|`simplify(expr)`   | Simplifies a mathematical expression |
|`factor(expr)`     | Factors an expression                |
|`expand(expr)`     | expand an expression                | 


---

## 🔹 Fractions

Operations on symbolic fractions.

| Keyword | Description |
| ---------------- | ----------------------------------- |
`numer(expr)` | Returns the numerator |
`denom(expr)` | Returns the denominator |
 

---

## 🔹 Exponentials

Operations with powers.

| Keyword | Description |
| ------------------------- | --------------------------------- |
`getbase(expr)`             | Base of a power |
`getexpo(expr)`             | Exponent |
`simplifybase(expr)`        | simplify Base of a power |
`simplifyexpo(expr)`        | simplify Exponent |
`expandbase(expr)`          | expand Base of a power |
`expandexpo(expr)`          | expand Exponent |
`factorse(expr)`            | factor Base of a power |
`factorexpo(expr)`          | factor Exponent |
